```python
# Usuário insere o primeiro valor numérico.
numero1 = int(input ("Digite o primeiro número:"))
# Usuário insere a operação matemática.
operacao = input("Digite a operação matemática (+, -, *, /): ")
# Usuário insere o segundo valor numérico.
numero2 = int(input ("Digite o segundo número:"))
# Usuário aciona o comando “Resultado”.
input("Pressione Enter para obter o resultado: ")
 # O sistema processa a operação e exibe o resultado.
while True:
    numero1 = float(input("Digite o primeiro número: "))
    operacao = input("Digite a operação (+, -, *, /): ")
    numero2 = float(input("Digite o segundo número: "))

    if operacao == "+":
        resultado = numero1 + numero2
    elif operacao == "-":
        resultado = numero1 - numero2
    elif operacao == "*":
        resultado = numero1 * numero2
    elif operacao == "/":
        resultado = numero1 / numero2
    else:
        print("Operação inválida")
        continue

    print("Resultado:", resultado)

    continuar = input("Deseja realizar outra operação? (s/n): ")

    if continuar == "n":
        break
```
